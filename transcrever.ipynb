```python
device = "cpu"
size = "large-v3-turbo"
```

```python
import whisper
from datetime import timedelta
```

```python
model = whisper.load_model(size, device=device)
result = model.transcribe(r"C:\Repositorios\Personal\Transcricao\Engenheiro_de_IA_2.mkv", 
                          language ='pt',
                          word_timestamps=True,
                          initial_prompt="Este vídeo é sobre uma entrevista de emprego sobre engenharia de IA em português.")

with open("transcricao_.srt", "w", encoding="utf-8") as f:
    for i, segment in enumerate(result["segments"], 1):
        start = timedelta(seconds=segment["start"])
        end = timedelta(seconds=segment["end"])
        text = segment["text"]
        f.write(f"{i}\n{start} --> {end}\n{text}\n\n")
```
